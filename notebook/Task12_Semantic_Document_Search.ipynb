## Loading ChromaDB

The ChromaDB database created in Task 11 was loaded and used as the foundation for semantic retrieval.

The database contained healthcare-related document embeddings generated using Sentence Transformers.

## Connecting LangChain with ChromaDB

LangChain was connected to the existing ChromaDB collection using the Chroma vector store interface.

The healthcare_documents collection was loaded for semantic document retrieval.

## Loading the Embedding Model

The model `all-MiniLM-L6-v2` from Sentence Transformers was used to convert user queries into vector embeddings.

This embedding model enables semantic similarity comparisons between queries and stored documents.

## Expanding the Knowledge Base

Ten additional healthcare documents were added to the existing ChromaDB collection.

The new documents covered:

- Mental Health
- Depression
- Anxiety Disorders
- Sleep
- Nutrition
- Exercise
- Cancer
- Electronic Health Records
- Wearable Devices
- Robotic Surgery

This expanded the knowledge base from 10 to 20 healthcare documents.

## Performing Semantic Search

User queries were read from `queries.csv`.

Each query was transformed into a vector embedding and compared against the embeddings stored in ChromaDB using semantic similarity search.

## Retrieving Top-5 Documents

For every query, the top 5 most relevant healthcare documents were retrieved.

Results were ranked according to similarity scores calculated by the vector database.

## Similarity Scores

Similarity scores were generated for all retrieved documents.

These scores indicate the semantic closeness between a user query and the retrieved healthcare documents.

The scores were stored in `similarity_scores.csv`.

## Comparison with Keyword Search

### Keyword Search

- Matches exact keywords.
- Limited understanding of context.

### Semantic Search

- Uses vector embeddings.
- Understands contextual meaning.
- Retrieves conceptually related documents.
- Produces more relevant search results.

### Observation

Semantic search demonstrated superior retrieval performance compared to traditional keyword-based search.

## Retrieval Analysis

Semantic search retrieves documents based on contextual meaning rather than exact keyword matching.

For example, a query such as **"What is diabetes?"** may retrieve not only the diabetes document but also related healthcare topics such as:

- Nutrition
- Hypertension
- Heart Disease
- Depression

This occurs because these topics are semantically related within the healthcare domain. Unlike keyword search, semantic search identifies conceptual relationships between documents and queries, leading to more meaningful retrieval results.
